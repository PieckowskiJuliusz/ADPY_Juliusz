```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. Generowanie losowych danych
np.random.seed(42)  # Opcjonalnie: zapewnia powtarzalność tych samych "losowych" wyników
x = np.arange(1, 51)  # Wartości na osi X (od 1 do 50)
y = np.random.randn(50).cumsum()  # Skumulowana suma losowych liczb (tworzy ciekawy trend)

# 2. Tworzenie wykresu
plt.figure(figsize=(10, 5))
plt.plot(x, y, marker='o', linestyle='-', color='#2b5c8f', label='Losowy przebieg')

# 3. Dodanie opisów i tytułu
plt.title('Wykres wygenerowany z losowych danych')
plt.xlabel('Numer pomiaru (X)')
plt.ylabel('Wartość (Y)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

# 4. Wyświetlenie wykresu
plt.tight_layout()
plt.show()
```
