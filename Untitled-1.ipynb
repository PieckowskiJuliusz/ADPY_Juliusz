```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

# losowanie 1000 liczb z rozkładu normalnego
dane = np.random.normal(loc=50, scale=10, size=1000)

# histogram
plt.hist(dane, bins=20, edgecolor='black')
plt.title("Histogram wylosowanych danych")
plt.xlabel("Wartość")
plt.ylabel("Liczność")
plt.show()
```
