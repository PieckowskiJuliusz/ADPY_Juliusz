```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(1, 21)
y = np.cumsum(np.random.randint(-5, 10, 20))

plt.plot(x, y, marker='o')
plt.title("Losowy trend")
plt.xlabel("Numer obserwacji")
plt.ylabel("Wartość")
plt.grid(True)
plt.show()
```
